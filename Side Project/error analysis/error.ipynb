```python
both_correct = cmp[(cmp.y_pred_16 == cmp.y_true) & (cmp.y_pred_19 == cmp.y_true)]
only_16_ok   = cmp[(cmp.y_pred_16 == cmp.y_true) & (cmp.y_pred_19 != cmp.y_true)]
only_19_ok   = cmp[(cmp.y_pred_16 != cmp.y_true) & (cmp.y_pred_19 == cmp.y_true)]
both_wrong   = cmp[(cmp.y_pred_16 != cmp.y_true) & (cmp.y_pred_19 != cmp.y_true)]

print("both_correct:", len(both_correct))
print("only_16_ok  :", len(only_16_ok))
print("only_19_ok  :", len(only_19_ok))
print("both_wrong  :", len(both_wrong))
```
